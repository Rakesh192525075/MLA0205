```python

# Experiment 4: ANN using Backpropagation Algorithm

import numpy as np

# Step 1: Create XOR dataset
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([
    [0],
    [1],
    [1],
    [0]
])

# Step 2: Initialize neural network
np.random.seed(42)

W1 = np.random.randn(2, 4)
b1 = np.zeros((1, 4))

W2 = np.random.randn(4, 1)
b2 = np.zeros((1, 1))

# Step 3: Sigmoid activation function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Step 4: Train using Backpropagation
learning_rate = 0.5
epochs = 10000

for i in range(epochs):

    # Forward propagation
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)

    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)

    # Backward propagation
    dz2 = a2 - y

    dw2 = np.dot(a1.T, dz2) / len(X)
    db2 = np.mean(dz2, axis=0, keepdims=True)

    dz1 = np.dot(dz2, W2.T) * a1 * (1 - a1)

    dw1 = np.dot(X.T, dz1) / len(X)
    db1 = np.mean(dz1, axis=0, keepdims=True)

    # Update weights and biases
    W2 -= learning_rate * dw2
    b2 -= learning_rate * db2

    W1 -= learning_rate * dw1
    b1 -= learning_rate * db1

# Step 5: Test the neural network
z1 = np.dot(X, W1) + b1
a1 = sigmoid(z1)

z2 = np.dot(a1, W2) + b2
predictions = sigmoid(z2)

# Step 6: Display results
predicted_output = (predictions >= 0.5).astype(int)

print("Actual Output:")
print(y.flatten())

print("\nPredicted Output:")
print(predicted_output.flatten())

accuracy = np.mean(predicted_output == y) * 100

print("\nAccuracy:", accuracy, "%")
```
